```python
import numpy as np
```

```python
data =np.genfromtxt('general_swim_extended_sine_fit.csv', delimiter=',', skip_header=1)
```

```python
from pathlib import Path

extended_data = np.empty((data.shape[0], 1 + 2 * (data.shape[1] - 1)), dtype=data.dtype)
extended_data[:, 0] = data[:, 0]
extended_data[:, 1::2] = data[:, 1:]
extended_data[:, 2::2] = 0

output_path = Path('/data/lilytorch/lilytorch/examples/pleurodeles/general_swim_extended_3D.csv')
np.savetxt(output_path, extended_data, delimiter=',')
print(f'Saved {extended_data.shape[1]} columns to {output_path}')
```
